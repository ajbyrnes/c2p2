```python
import os
import pandas as pd

import plotly.express as px
import plotly.graph_objects as go
```

```python
results_dir = "csv"

results_df = pd.DataFrame()

for file in os.listdir(results_dir):
    temp_df = pd.read_csv(os.path.join(results_dir, file))
    temp_df["resultsFile"] = file
    results_df = pd.concat([results_df, temp_df], ignore_index=True)
    
# Drop empty columns
results_df = results_df.dropna(axis=1, how='all')

# Add compressor name
results_df['compressorName'] = results_df['compressor'].str.split('{').str[0]

# Add input size in MB
results_df['inputSizeMB'] = results_df['inputSizeBytes'] / (1024 * 1024)

# Remap data name
data_name_map = {
    'AnalysisJetsAuxDyn.pt': 'pt',
    'AnalysisJetsAuxDyn.eta': 'eta',
    'AnalysisJetsAuxDyn.phi': 'phi',
}

results_df['dataName'] = results_df['dataName'].map(data_name_map).fillna(results_df['dataName'])

# Remap compression algorithm name
#  0 = ALGO_LORENZO_REG
#  1 = ALGO_INTERP_LORENZO
#  2 = ALGO_INTERP
#  3 = ALGO_NOPRED
#  4 = ALGO_LOSSLESS

algorithm_name_map = {
    0: "Lorenzo Predictor (Regression)",
    1: "Lorenzo Predictor (Interpolation)",
    2: "Interpolation",
    3: "No Predictor",
    4: "Lossless Compression"
}

results_df['algorithmName'] = results_df['algorithm'].map(algorithm_name_map).fillna(results_df['algorithm'])

# Add downgraded column
results_df['downgraded'] = results_df['maxAbsError'] == 0

# Add combination compressor-branch
results_df['compressorBranch'] = results_df['compressorName'] + '-' + results_df['dataName']

results_df
```

```python
plot_df = results_df[
    (results_df['meanRelErrorPct'] <= 1)
]

fig = px.scatter(
    plot_df,
    x='inputSizeMB',
    y='compressionRatio',
    color='compressorBranch',
    size='meanRelErrorPct',
    hover_data=['compressorName', 'dataName', 'resultsFile'],
    title='Compression Ratio vs Input Size with Mean Relative Error Size',
    labels={
        'inputSizeMB': 'Input Size (MB)',
        'compressionRatio': 'Compression Ratio',
        'meanRelErrorPct': 'Mean Relative Error (%)'
    }
)

fig.show()
```

```python
# Drop downgraded rows
good_df = results_df[~results_df['downgraded']]

# Drop rows with error >1%
good_df = good_df[good_df['meanRelErrorPct'] <= 1]

good_sz3 = good_df[good_df['compressorName'] == 'SZ3Compressor']
good_trunc = good_df[good_df['compressorName'] == 'TruncCompressor']
```

```python
colors = [
    "#0b80c3",  # ATLAS blue
    "#c34e0b",  # Complementary
    "#c30b80",  # Triadic 1
    "#80c30b",  # Triadic 2
    "#0b24c3",  # Analogous 1
    "#0bc3aa",  # Analogous 2
    "#c30b24",  # Split 1
    "#c3aa0b",  # Split 2
    "#09669c",  # Monochromatic 1
    "#0d9aea"   # Monochromatic 2
]

# Create continuous color scale
blue_low = px.colors.make_colorscale(colors[:2])
blue_high = px.colors.make_colorscale(colors[:2][::-1])
```

```python
# Heatmap of compression ratios
heatmap_df = good_sz3.pivot_table(
    index='algorithmName',
    columns='dataName',
    values='compressionRatio',
    aggfunc='mean'
).fillna(0)

heatmap_fig = px.imshow(
    heatmap_df,
    labels=dict(x="Data Name", y="Compressor", color="Compression Ratio"),
    x=heatmap_df.columns,
    y=heatmap_df.index,
    color_continuous_scale=blue_high,
    title="Average Compression Ratios by Compressor and Data Name"
)

heatmap_fig.update_layout(
    xaxis_title="Data Name",
    yaxis_title="Compressor",
    coloraxis_colorbar=dict(title="Compression Ratio"),
    width=800,
    height=600
)
```
